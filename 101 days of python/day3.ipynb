## In Python, a module is a file containing Python code, such as functions, variables, and classes, that you can reuse in other programs. A Python module usually has a .py extension.
## Types of Modules in python 
There are 3 main types of python modules:
1.Built-in modules
These modules are included with Python, so you can use them without installing addtional packages.
Examples: a.math b.random c.os d.sys e.datetime
purposes respectively are as follows:
->mathematical operations
->generates random values 
->interacts with the operating system 
->provides access to python interpreter features
->Works with dates and times


Example:
import math
print(math.sqrt(25))
print(math.pi)
